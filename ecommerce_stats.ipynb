
# Statistics Practice Tasks

This notebook contains **20 simple statistics tasks** based on an **ecommerce dataset**.

Dataset file used:
`ecommerce_dataset_1000_rows.csv`

Topics covered:
- Mean, median, mode
- Min, max, range
- Variance and standard deviation
- Quartiles and IQR
- Skewness and kurtosis
- Correlation and covariance
- Grouped statistical summaries


```python

import pandas as pd

df = pd.read_csv("/mnt/data/ecommerce_dataset_1000_rows.csv")
df.head()
```

## Task 1

Find the mean of Customer_Age.

```python
df['Customer_Age'].mean()
```

## Task 2

Find the median of Customer_Age.

```python
df['Customer_Age'].median()
```

## Task 3

Find the mode of Product_Category.

```python
df['Product_Category'].mode()
```

## Task 4

Find the mean of Unit_Price.

```python
df['Unit_Price'].mean()
```

## Task 5

Find the median of Final_Amount.

```python
df['Final_Amount'].median()
```

## Task 6

Find the minimum and maximum of Quantity.

```python
df['Quantity'].min(), df['Quantity'].max()
```

## Task 7

Find the range of Unit_Price.

```python
df['Unit_Price'].max() - df['Unit_Price'].min()
```

## Task 8

Find the variance of Final_Amount.

```python
df['Final_Amount'].var()
```

## Task 9

Find the standard deviation of Final_Amount.

```python
df['Final_Amount'].std()
```

## Task 10

Find the first quartile (Q1) of Unit_Price.

```python
df['Unit_Price'].quantile(0.25)
```

## Task 11

Find the third quartile (Q3) of Unit_Price.

```python
df['Unit_Price'].quantile(0.75)
```

## Task 12

Find the interquartile range (IQR) of Unit_Price.

```python
df['Unit_Price'].quantile(0.75) - df['Unit_Price'].quantile(0.25)
```

## Task 13

Find the skewness of Final_Amount.

```python
df['Final_Amount'].skew()
```

## Task 14

Find the kurtosis of Final_Amount.

```python
df['Final_Amount'].kurt()
```

## Task 15

Find the correlation between Quantity and Final_Amount.

```python
df['Quantity'].corr(df['Final_Amount'])
```

## Task 16

Find the covariance between Unit_Price and Final_Amount.

```python
df['Unit_Price'].cov(df['Final_Amount'])
```

## Task 17

Calculate the mean Final_Amount for each Product_Category.

```python
df.groupby('Product_Category')['Final_Amount'].mean()
```

## Task 18

Calculate the median Unit_Price for each Customer_Segment.

```python
df.groupby('Customer_Segment')['Unit_Price'].median()
```

## Task 19

Find how many orders are above the average Final_Amount.

```python
avg_final = df['Final_Amount'].mean()
df[df['Final_Amount'] > avg_final].shape[0]
```

## Task 20

Create a summary table for Customer_Age, Quantity, Unit_Price, Discount_Percent, and Final_Amount.

```python
df[['Customer_Age', 'Quantity', 'Unit_Price', 'Discount_Percent', 'Final_Amount']].describe()
```
