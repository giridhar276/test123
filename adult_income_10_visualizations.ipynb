
# 10 Visualizations Based on the Uploaded CSV

This notebook uses the uploaded dataset and creates **10 simple visualizations** with **matplotlib** and **seaborn**.


```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"/mnt/data/d954a738-d102-49c9-ac7b-2b1a761b4261.csv")
df.head()
```

## Visualization 1: Income Distribution

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='income')
plt.title('Income Distribution')
plt.show()
```

## Visualization 2: Age Distribution

```python
plt.figure(figsize=(7,4))
plt.hist(df['age'], bins=20)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

## Visualization 3: Workclass Distribution

```python
plt.figure(figsize=(8,4))
sns.countplot(data=df, x='workclass')
plt.title('Workclass Distribution')
plt.xticks(rotation=45)
plt.show()
```

## Visualization 4: Education Distribution

```python
plt.figure(figsize=(10,4))
sns.countplot(data=df, x='education', order=df['education'].value_counts().index)
plt.title('Education Distribution')
plt.xticks(rotation=45)
plt.show()
```

## Visualization 5: Gender Distribution

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='gender')
plt.title('Gender Distribution')
plt.show()
```

## Visualization 6: Hours per Week Distribution

```python
plt.figure(figsize=(7,4))
sns.histplot(df['hours-per-week'], bins=20, kde=True)
plt.title('Hours per Week Distribution')
plt.show()
```

## Visualization 7: Age vs Hours per Week

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=df, x='age', y='hours-per-week')
plt.title('Age vs Hours per Week')
plt.show()
```

## Visualization 8: Income by Gender

```python
plt.figure(figsize=(7,4))
sns.countplot(data=df, x='gender', hue='income')
plt.title('Income by Gender')
plt.show()
```

## Visualization 9: Average Hours per Week by Income

```python
avg_hours = df.groupby('income')['hours-per-week'].mean().reset_index()
plt.figure(figsize=(6,4))
sns.barplot(data=avg_hours, x='income', y='hours-per-week')
plt.title('Average Hours per Week by Income')
plt.show()
```

## Visualization 10: Correlation Heatmap

```python
num_df = df[['age', 'fnlwgt', 'educational-num', 'capital-gain', 'capital-loss', 'hours-per-week']]
corr = num_df.corr()
plt.figure(figsize=(8,5))
sns.heatmap(corr, annot=True, cmap='Blues')
plt.title('Correlation Heatmap')
plt.show()
```
